```python
X = df.drop(columns= ['Species','Id'], axis=1)
y = df['Species']
```

```python
encoder = LabelEncoder()
y_int = encoder.fit_transform(y)
```

```python
y_int
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y_int, test_size=0.2, random_state=42, stratify=y_int)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
per = Perceptron(max_iter=1000, random_state=42) # forward and backward iteration is called max iteration means that :
```

### Perceptron

per = Perceptron(max_iter=1000, random_state=42)

- `Perceptron` → Creates a Perceptron classification model.

- `max_iter=1000` → Specifies the maximum number of iterations (epochs) during training.
  - In each iteration, the model goes through the training data.
  - It makes predictions (forward pass).
  - It checks the prediction error.
  - It updates the weights for misclassified samples.
  - This process is repeated for a maximum of 1000 complete passes over the training data.
  - The model can stop earlier if it converges.

- `random_state=42` → Makes the training process reproducible, so we get the same result each time we run the code.

> Note: `max_iter=1000` does not mean 1000 forward passes + 1000 backward passes separately. It means up to 1000 complete epochs, where predictions and weight updates happen during each epoch.

```python
per.fit(X_train_scaled, y_train)
```

```python
y_pred_percep = per.predict(X_test_scaled)
```

```python
accuracy_score(y_test, y_pred_percep)
```

```python
print(classification_report(y_test, y_pred_percep))
```

```python
y_train_cat = to_categorical(y_train, num_classes=3)
y_test_cat = to_categorical(y_test,num_classes=3)
```

### Converting Labels to Categorical Format

y_train_cat = to_categorical(y_train, num_classes=3)

y_test_cat = to_categorical(y_test, num_classes=3)

- `to_categorical()` → Converts class labels into **one-hot encoded vectors**.

- `num_classes=3` → We have **3 classes** in our classification problem.

For example, if:

y_train = [0, 1, 2, 1, 0]

After applying `to_categorical()`:

y_train_cat = [
    [1, 0, 0],   # Class 0
    [0, 1, 0],   # Class 1
    [0, 0, 1],   # Class 2
    [0, 1, 0],   # Class 1
    [1, 0, 0]    # Class 0
]

### Why do we need this?

For a neural network with **3 output classes**, the output layer usually has 3 neurons:

model.add(Dense(3, activation='softmax'))

The `softmax` layer produces probabilities for each class, for example:

[0.05, 0.90, 0.05]

The actual target should therefore be in the same 3-value format:

[0, 1, 0]

So, `to_categorical()` converts:

Class label → One-hot vector

0 → [1, 0, 0]
1 → [0, 1, 0]
2 → [0, 0, 1]

This makes the target labels compatible with the neural network's 3-output softmax layer and categorical cross-entropy loss.

- `y_train_cat` → One-hot encoded training labels.
- `y_test_cat` → One-hot encoded testing labels.

```python
model = Sequential([
    Dense(16, activation='relu', input_dim=4),
    Dense(8, activation='relu'),
    Dense(3, activation='softmax') # when multi class
])
```

```python
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
history = model.fit(X_train_scaled, y_train_cat, epochs=100, batch_size=8, validation_split=0.2, verbose=1)
```

```python
loss, acc = model.evaluate(X_test_scaled, y_test_cat,verbose=0)
print(f"Test Loss: {loss:.4f}, Test Accuracy: {acc*100:.2f}%")
```

```python
plt.figure(figsize=(10,4))
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.show()
```
